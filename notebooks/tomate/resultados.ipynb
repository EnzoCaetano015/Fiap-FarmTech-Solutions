Resultados reais do experimento PlantDoc. Esta leitura não inicia treinamento. Veja docs/colab.md para reprodução.

```
from pathlib import Path
import json
candidates = [Path.cwd(), *Path.cwd().parents]
BASE = next((p for p in candidates if (p / "models/tomate/manifest.json").is_file()), None)
if BASE is None:
    raise RuntimeError("Bundle agrícola ausente: extraia-o em models/tomate na raiz do projeto")
manifest = json.loads((BASE / "models/tomate/manifest.json").read_text(encoding="utf-8"))
selection = json.loads((BASE / "models/tomate/selection.json").read_text(encoding="utf-8"))
print(json.dumps({"epocas_selecionadas": selection["selected_epochs"], "validacao": [{k:r[k] for k in ("epochs","map50","map50_95","training_seconds")} for r in selection["runs"]]}, indent=2))
```
